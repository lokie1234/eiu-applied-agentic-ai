##**Structured vs. Unstructured Data**


image.png

To do analysis, data needs to be **structured as a table with rows and columns**.

Most data sources are already organised in this way and are ready for analysis. Unstructured data sources require additional transformations before doing analysis.

To do analysis in Python, data must be reside within a Pandas DataFrame.
* DataFrames look just like regular tables with rows and columns, but with additional features.
* External structured and semi-structured data can be read directly into a DataFrame.

##**Reading Data Into Python**
Pandas allows you to read in data from mutiple file formats.
image.png

###**1. Read CSV Files**
Use the `pd.read_csv()` to read flat files by specifying the file path.


image.png

*Load Data: student_data.csv*

```python
# Import Pandas
import pandas as pd
```

```python
# Read data file
df = pd.read_csv("student_data.csv")
```

###**2. Read Excel Files**

Use the `pd.read_excel()` to read Excel files by specifying the file path.
* You can use the "sheet_name" argument to specify the worksheet (the default is 0 - first sheet)

*Load Data: Course Offerings.xlsx*

```python
# Read data file
df2 = pd.read_excel("Course Offerings.xlsx")
```

```python
df2.head()
```

```python
# Use sheet_name argument to access Excel file with mutiple worksheet
df2 = pd.read_excel("Course Offerings.xlsx", sheet_name=1)
```

```python
df2.head()
```

##**Quickly Explore a DataFrame**

```python
# Shows the first 5 rows of the DataFrame
df.head()
```
